```python
import os
import glob
from PIL import Image
import matplotlib.pyplot as plt
```

```python
hr_dir = "../data/raw/DIV2K/DIV2K_train_HR"
lr_dir_x2 = "../data/processed/DIV2K/DIV2K_train_LR_x2"
lr_dir_x4 = "../data/processed/DIV2K/DIV2K_train_LR_x4"
lr_dir_x8 = "../data/processed/DIV2K/DIV2K_train_LR_x8"
```

```python
hr_images = sorted(glob.glob(os.path.join(hr_dir, "*.png")))
lr_images_x2 = sorted(glob.glob(os.path.join(lr_dir_x2, "*.png")))
lr_images_x4 = sorted(glob.glob(os.path.join(lr_dir_x4, "*.png")))
lr_images_x8 = sorted(glob.glob(os.path.join(lr_dir_x8, "*.png")))
print("Number of images HR :", len(hr_images))
```

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 15))
idx = 2
img_hr = Image.open(hr_images[idx])
axes[0][0].imshow(img_hr)
axes[0][0].set_title('HR Image')
axes[0][0].axis('off')

img_lr_x2 = Image.open(lr_images_x2[idx])
axes[0][1].imshow(img_lr_x2)
axes[0][1].set_title('LR x2 Image')
axes[0][1].axis('off')

img_lr_x4 = Image.open(lr_images_x4[idx])
axes[1][0].imshow(img_lr_x4)
axes[1][0].set_title('LR x4 Image')
axes[1][0].axis('off')

img_lr_x8 = Image.open(lr_images_x8[idx])
axes[1][1].imshow(img_lr_x8)
axes[1][1].set_title('LR x8 Image')
axes[1][1].axis('off')

plt.tight_layout()
plt.show()
```
